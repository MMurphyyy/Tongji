# Tensorflow2.0 小练习

```python
import tensorflow as tf
import numpy as np
tf.config.list_physical_devices('GPU')
```

## 实现softmax函数

```python
def softmax(x):
    ##########
    '''实现softmax函数，只要求对最后一维归一化，
    不允许用tf自带的softmax函数'''
    ##########
     # 计算指数值
    exp_x = np.exp(x - np.max(x, axis=-1, keepdims=True))  # 减去最大值以提高数值稳定性
    # 对最后一维进行归一化
    prob_x = exp_x / np.sum(exp_x, axis=-1, keepdims=True)
    return prob_x

# 测试数据
test_data = np.random.normal(size=[10, 5])

# 计算自定义 softmax 和 TensorFlow 的 softmax
custom_softmax = softmax(test_data)
tf_softmax = tf.nn.softmax(test_data, axis=-1).numpy()

# 检查两者是否接近
assert np.all((custom_softmax - tf_softmax)**2 < 0.0001), "自定义 softmax 与 TensorFlow 的 softmax 不一致"

print("自定义 softmax 与 TensorFlow 的 softmax 一致")
```

## 实现sigmoid函数

```python
def sigmoid(x):
    ##########
    '''实现sigmoid函数， 不允许用tf自带的sigmoid函数'''
    ##########
    prob_x = 1 / (1 + np.exp(-x))
    return prob_x

# 测试数据
test_data = np.random.normal(size=[10, 5])

# 计算自定义 sigmoid 和 TensorFlow 的 sigmoid
custom_sigmoid = sigmoid(test_data)
tf_sigmoid = tf.nn.sigmoid(test_data).numpy()

# 检查两者是否接近
assert np.all((custom_sigmoid - tf_sigmoid)**2 < 0.0001), "自定义 sigmoid 与 TensorFlow 的 sigmoid 不一致"

print("自定义 sigmoid 与 TensorFlow 的 sigmoid 一致")
```

## 实现 softmax 交叉熵loss函数

```python
def softmax_ce(x, label):
    # 计算交叉熵损失
    loss = -np.sum(label * np.log(x + 1e-9), axis=-1)  # 加上 1e-9 避免 log(0) 的情况
    loss = np.mean(loss)  # 对批量数据取平均
    return loss

# 测试数据
test_data = np.random.normal(size=[10, 5])
prob = tf.nn.softmax(test_data).numpy()  # 计算 softmax 输出
label = np.zeros_like(test_data)
label[np.arange(10), np.random.randint(0, 5, size=10)] = 1.  # 生成随机 one-hot 标签

# 计算自定义 softmax 交叉熵损失和 TensorFlow 的损失
custom_loss = softmax_ce(prob, label)
tf_loss = tf.reduce_mean(tf.nn.softmax_cross_entropy_with_logits(label, test_data)).numpy()

# 检查两者是否接近
assert (custom_loss - tf_loss)**2 < 0.0001, "自定义 softmax 交叉熵损失与 TensorFlow 的损失不一致"

print("自定义 softmax 交叉熵损失与 TensorFlow 的损失一致")
```

## 实现 sigmoid 交叉熵loss函数

```python
def sigmoid_ce(x, label):
    # 计算二分类交叉熵损失
    loss = -(label * np.log(x + 1e-9) + (1 - label) * np.log(1 - x + 1e-9))  # 加上 1e-9 避免 log(0) 的情况
    loss = np.mean(loss)  # 对批量数据取平均
    return loss

# 测试数据
test_data = np.random.normal(size=[10])
prob = tf.nn.sigmoid(test_data).numpy()  # 计算 sigmoid 输出
label = np.random.randint(0, 2, 10).astype(test_data.dtype)  # 生成随机二分类标签
print("真实标签:", label)

# 计算自定义 sigmoid 交叉熵损失和 TensorFlow 的损失
custom_loss = sigmoid_ce(prob, label)
tf_loss = tf.reduce_mean(tf.nn.sigmoid_cross_entropy_with_logits(label, test_data)).numpy()

# 检查两者是否接近
assert (custom_loss - tf_loss)**2 < 0.0001, "自定义 sigmoid 交叉熵损失与 TensorFlow 的损失不一致"

print("自定义 sigmoid 交叉熵损失与 TensorFlow 的损失一致")
```
